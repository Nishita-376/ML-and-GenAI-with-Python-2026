```python
#ques 1
import pandas as pd

df = pd.read_csv('agriculture_ml_dataset_cw.csv')

print("Rows and Columns:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 10 Records:")
display(df.head(10))

['rainfall_mm',
 'temperature_c',
 'fertilizer_kg',
 'irrigation_hours',
 'soil_ph',
 'crop_type',
 'soil_type',
 'yield_ton_per_hectare']
```

```python
#ques 2
print("Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#ques 3
stats = df.describe()
display(stats)

print("\nHighest Mean Feature:")
print(stats.loc['mean'].idxmax())

print("\nHighest Standard Deviation Feature:")
print(stats.loc['std'].idxmax())
```

```python
#ques 4
import matplotlib.pyplot as plt

columns = [
    'Rainfall',
    'Temperature',
    'Fertilizer',
    'Yield'
]

for col in columns:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=20)
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
#ques 5
import seaborn as sns
import matplotlib.pyplot as plt

print(df['Crop_Type'].value_counts())

plt.figure(figsize=(6,4))
sns.countplot(x='Crop_Type', data=df)
plt.title("Crop Type Count")
plt.show()
```

```python
#ques 6
print(df['Soil_Type'].value_counts())

plt.figure(figsize=(6,4))
sns.countplot(x='Soil_Type', data=df)
plt.title("Soil Type Count")
plt.show()
```

```python
#ques 7
plt.figure(figsize=(6,4))
plt.hist(df['Yield'], bins=20)
plt.title('Yield Distribution')
plt.xlabel('Yield')
plt.ylabel('Frequency')
plt.show()
```

```python
#ques 8
plt.figure(figsize=(6,4))
plt.scatter(df['Rainfall'], df['Yield'])
plt.xlabel('Rainfall')
plt.ylabel('Yield')
plt.title('Rainfall vs Yield')
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df['Fertilizer'], df['Yield'])
plt.xlabel('Fertilizer')
plt.ylabel('Yield')
plt.title('Fertilizer vs Yield')
plt.show()
```

```python
#ques 9
numeric_df = df.select_dtypes(include='number')

corr_matrix = numeric_df.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix,
            annot=True,
            cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()

print(corr_matrix['Yield'].sort_values(ascending=False))
```

```python
#ques 10
crop_avg = df.groupby('Crop_Type')['Yield'].mean()
soil_avg = df.groupby('Soil_Type')['Yield'].mean()

print("Average Yield by Crop")
print(crop_avg)

print("\nAverage Yield by Soil")
print(soil_avg)
```

```python
# ques 11
encoded_df = pd.get_dummies(
    df,
    columns=['Crop_Type', 'Soil_Type'],
    drop_first=True
)

display(encoded_df.head())
```

```python
#ques 12
X = encoded_df.drop('Yield', axis=1)
y = encoded_df['Yield']

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
#ques 13
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# ques 14
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Intercept:")
print(model.intercept_)

coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})

display(coef_df.sort_values(
    by='Coefficient',
    ascending=False
))

print("\nHighest Positive Coefficient Feature:")
print(
    coef_df.loc[
        coef_df['Coefficient'].idxmax(),
        'Feature'
    ]
)
```
